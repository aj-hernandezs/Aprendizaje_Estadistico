Para los datos de los archivos "bateria_tableta.txt" y "Salarios_minimos.csv":

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import numpy as np
```

#### Dataset: bateria_tableta

```python
bt = pd.read_csv('Datasets/bateria_tableta.txt')
```

```python
# Orden 1
x_v = np.reshape(bt['carga'], [-1, 1])
arr1 = x_v
lr = LinearRegression()
lr.fit(arr1, bt['duracion'])
bt['pred_1'] = lr.predict(arr1)
print(f'R2 Orden 1: {lr.score(arr1, bt["duracion"])}')

# Orden 5
x2_v = x_v**2
x3_v = x_v**3
x4_v = x_v**4
x5_v = x_v**5
arr2 = np.hstack([x_v, x2_v, x3_v, x4_v, x5_v])
lr.fit(arr2, bt['duracion'])
bt['pred_2'] = lr.predict(arr2)
print(f'R2 Orden 5: {lr.score(arr2, bt["duracion"])}')
# Orden 9
x6_v = x_v**6
x7_v = x_v**7
x8_v = x_v**8
x9_v = x_v**9
arr3 = np.hstack([x_v, x2_v, x3_v, x4_v, x5_v, x6_v, x7_v, x8_v, x9_v])
lr.fit(arr3, bt['duracion'])
bt['pred_3'] = lr.predict(arr3)
print(f'R2 Orden 9: {lr.score(arr3, bt["duracion"])}')

#Graficas
plt.figure(figsize=(10, 6))
plt.scatter( bt['carga'], bt['duracion'], label='x',color='black', edgecolors='white')
plt.scatter(bt['carga'], bt['pred_1'], label='Orden 1')
plt.scatter(bt['carga'], bt['pred_2'], label='Orden 5')
plt.scatter(bt['carga'], bt['pred_3'], label='Orden 9')
plt.title('Duración vs Carga')
plt.xlabel('Carga')
plt.ylabel('Duración')
plt.legend()
plt.show()
```
