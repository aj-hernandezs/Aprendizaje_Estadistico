## Ejercicio de Codificación

### Carga de Dataframe

```python
import pandas as pd

df = pd.read_excel('Datasets/Aguas_el_feo.xlsx', sheet_name='Sheet1', header=0)

df
```

### Renombramiento de observaciones

Las variables categóricas de tipo ordinal y booleanas al tener un orden y ser binarias respectivamente, se les puede asignar un valor númerico que respeten ese orden para las que tienen orden y asignar 0-1 para las booleanas.

```python
df['Tamaño'] = df['Tamaño'].map({'chica': 0, 'grande': 1})
df['Hielo'] = df['Hielo'].map({'no': 0, 'si': 1})
df
```

Para variables no ordinales se crea una columna por cada valor único dandole valor de 1 a la fila en la que pertenecía y 0 el resto

```python
df['Naranja'] = df["Sabor"].str.contains("naranja")
df['Coco'] = df["Sabor"].str.contains("coco")
df['Kahlua'] = df["Sabor"].str.contains("kahlua")
df['FresaPiña'] = df["Sabor"].str.contains("fresapiña")
df['Sandia'] = df["Sabor"].str.contains("sandia")
df = df.drop(columns=['Sabor'])
```

### Regresión Multiple con StastsModels / OLS

```python
import statsmodels.api as sm

df[['Naranja', 'Coco', 'Kahlua', 'FresaPiña', 'Sandia']] = df[['Naranja', 'Coco', 'Kahlua', 'FresaPiña', 'Sandia']].astype(int)
X = df[['Tamaño', 'Hielo', 'Naranja', 'Coco', 'Kahlua', 'FresaPiña', 'Sandia']]
X = sm.add_constant(X)
Y = df['Calificacion']

ols = sm.OLS(Y, X).fit()
ols.summary()
```

### Uso de SKlearn para transformación de variables (Ingeniería de Características)

```python
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder

df_sk = pd.read_excel('Datasets/Aguas_el_feo.xlsx', sheet_name='Sheet1', header=0)

enc = OrdinalEncoder(categories=[['chica', 'grande'], ['no', 'si']])
df_sk[['Tamaño', 'Hielo']] = enc.fit_transform(df_sk[['Tamaño', 'Hielo']])

one_hot_enc = OneHotEncoder(categories=[['naranja', 'coco', 'kahlua', 'fresapiña', 'sandia']], sparse_output=False)
df_sk[['Naranja','Coco','Kahlua','FresaPiña','Sandia']] = one_hot_enc.fit_transform(df_sk[['Sabor']])
df_sk.drop(columns=['Sabor'], inplace=True)
df_sk

```

### Regresión de Comprobación

```python
X = df_sk[['Tamaño', 'Hielo', 'Naranja', 'Coco', 'Kahlua', 'FresaPiña', 'Sandia']]
X = sm.add_constant(X)
Y = df_sk['Calificacion']
ols_sk = sm.OLS(Y, X).fit()
ols_sk.summary()
```

Los resultados usando ambas maneras para la transformación de variables son exactamente iguales.
