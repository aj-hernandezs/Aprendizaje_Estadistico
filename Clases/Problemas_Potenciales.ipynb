## Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('Datasets/Advertising.csv')
```

## Regresión Multiple

```python
import statsmodels.api as sm
X = df[['TV', 'radio', 'newspaper']]
y = df['sales']
X = sm.add_constant(X)
OLS_model = sm.OLS(y, X).fit()
OLS_model.summary()
```
