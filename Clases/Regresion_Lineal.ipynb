## Regresión Lineal

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('Datasets/bateria_tableta.txt')
print(df.columns)
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df['carga'], df['duracion'])
plt.xlabel('Carga')
plt.ylabel('Duración')
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit?
```

```python
import numpy as np

x_v = np.reshape(df['carga'], [-1, 1])
x2_v = x_v**2
ones =np.ones([100,1])
```

```python
arr =np.hstack([x_v, x2_v])
lr.fit(arr, df['duracion'])
print(lr.intercept_)
```

```python
print(lr.coef_)
```

B0 = lr.intercept_
B1..n = lr.coef_ (Segun el orden de la matriz X)

```python
df['pred'] = lr.intercept_ + lr.coef_[0]*df['carga'] + lr.coef_[1]*df['carga']**2
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df['carga'], df['duracion'])
plt.scatter(df['carga'], df['pred'])
plt.show()
```

```python
pred =lr.predict(arr)
plt.figure(figsize=(10, 6))
plt.scatter(df['carga'], pred)
```
