# Descripcion de las variables

El archivo `StudentPerformanceFactors.csv` contiene **6,607 registros** de estudiantes y **20 variables**. La variable de resultado es `Exam_Score`; las demas describen habitos, contexto familiar, caracteristicas escolares y factores personales.

| Variable | Tipo | Descripcion y valores |
|---|---|---|
| `Hours_Studied` | Numerica discreta | Horas dedicadas al estudio. |
| `Attendance` | Numerica continua | Porcentaje de asistencia del estudiante. |
| `Parental_Involvement` | Categorica ordinal | Nivel de participacion de los padres: `Low`, `Medium` o `High`. |
| `Access_to_Resources` | Categorica ordinal | Nivel de acceso a recursos educativos: `Low`, `Medium` o `High`. |
| `Extracurricular_Activities` | Categorica binaria | Participacion en actividades extracurriculares: `Yes` o `No`. |
| `Sleep_Hours` | Numerica discreta | Horas de sueno por dia. |
| `Previous_Scores` | Numerica continua | Calificacion obtenida previamente. |
| `Motivation_Level` | Categorica ordinal | Nivel de motivacion: `Low`, `Medium` o `High`. |
| `Internet_Access` | Categorica binaria | Disponibilidad de acceso a internet: `Yes` o `No`. |
| `Tutoring_Sessions` | Numerica discreta | Numero de sesiones de tutoria recibidas. |
| `Family_Income` | Categorica ordinal | Nivel de ingreso familiar: `Low`, `Medium` o `High`. |
| `Teacher_Quality` | Categorica ordinal | Calidad percibida del docente: `Low`, `Medium` o `High`. |
| `School_Type` | Categorica nominal | Tipo de escuela: `Public` o `Private`. |
| `Peer_Influence` | Categorica ordinal | Influencia de los pares: `Negative`, `Neutral` o `Positive`. |
| `Physical_Activity` | Numerica discreta | Horas dedicadas a actividad fisica. |
| `Learning_Disabilities` | Categorica binaria | Indica la presencia de dificultades de aprendizaje: `Yes` o `No`. |
| `Parental_Education_Level` | Categorica ordinal | Maximo nivel educativo de los padres: `High School`, `College` o `Postgraduate`. |
| `Distance_from_Home` | Categorica ordinal | Distancia entre el hogar y la escuela: `Near`, `Moderate` o `Far`. |
| `Gender` | Categorica nominal | Genero registrado: `Female` o `Male`. |
| `Exam_Score` | Numerica continua | Calificacion del examen; es la variable objetivo para explicar o predecir el desempeno academico. |

> **Nota de calidad de datos:** el archivo contiene valores faltantes en varias columnas, incluidos valores codificados como `NA`. Conviene convertirlos a valores nulos y revisar los valores atipicos antes del analisis.
